```python
invoice_dates = pd.to_datetime(df["InvoiceDate"])

print("Earliest transaction:", invoice_dates.min())
print("Latest transaction:", invoice_dates.max())
```

```python
print("Number of unique months:",
      invoice_dates.dt.to_period("M").nunique())
```

```python
print("Months covered:")
print(sorted(invoice_dates.dt.to_period("M").unique()))
```

```python
monthly_transactions = (
    invoice_dates.dt.to_period("M")
    .value_counts()
    .sort_index()
)

monthly_transactions
```

```python
invoice_numbers = df["InvoiceNo"].astype(str)

cancelled_mask = invoice_numbers.str.startswith("C")

print("Cancelled transaction rows:", cancelled_mask.sum())
```

```python
cancelled_percentage = (cancelled_mask.sum() / len(df)) * 100

print("Percentage of cancelled transaction rows:",
      cancelled_percentage)
```

```python
cancelled_invoices = df.loc[cancelled_mask, "InvoiceNo"].nunique()

print("Unique cancelled invoices:", cancelled_invoices)
```

```python
invoice_type_counts = (
    df["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
    .value_counts()
)

print(invoice_type_counts)
```

```python
customer_missing_cancelled = pd.crosstab(
    df["CustomerID"].isna(),
    cancelled_mask
)

customer_missing_cancelled
```

```python
missing_description = df["Description"].isna()

print("Rows with missing Description:", missing_description.sum())
print(
    "Missing Description + Missing CustomerID:",
    (missing_description & df["CustomerID"].isna()).sum()
)
print(
    "Missing Description + Cancelled:",
    (missing_description & cancelled_mask).sum()
)
```

```python
country_summary = (
    df["Country"]
    .value_counts()
    .rename_axis("Country")
    .reset_index(name="Transactions")
)

country_summary.head(15)
```

```python
print("Transactions from United Kingdom:",
      (df["Country"] == "United Kingdom").sum())

print(
    "Percentage from United Kingdom:",
    (df["Country"] == "United Kingdom").mean() * 100
)
```

```python
customer_transaction_counts = (
    df["CustomerID"]
    .value_counts()
)

print("Minimum transactions by identifiable customer:",
      customer_transaction_counts.min())

print("Median transactions per customer:",
      customer_transaction_counts.median())

print("Maximum transactions by customer:",
      customer_transaction_counts.max())
```
