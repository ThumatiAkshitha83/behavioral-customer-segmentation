```python
# Step 1: Remove exact duplicate transaction rows

before_duplicates = len(clean_df)

clean_df = clean_df.drop_duplicates().copy()

after_duplicates = len(clean_df)
duplicates_removed = before_duplicates - after_duplicates

print("Rows before:", before_duplicates)
print("Rows after:", after_duplicates)
print("Duplicate rows removed:", duplicates_removed)
```

```python
# Step 2: Remove transactions without CustomerID

before_customerid = len(clean_df)

missing_customerid = clean_df["CustomerID"].isna().sum()

clean_df = clean_df[
    clean_df["CustomerID"].notna()
].copy()

after_customerid = len(clean_df)

print("Rows before:", before_customerid)
print("Missing CustomerID rows removed:", missing_customerid)
print("Rows after:", after_customerid)
print("Customers remaining:", clean_df["CustomerID"].nunique())
```

```python
# Step 3: Remove cancelled invoices

before_cancelled = len(clean_df)

cancelled_mask = clean_df["InvoiceNo"].astype(str).str.startswith("C")
cancelled_removed = cancelled_mask.sum()

clean_df = clean_df[
    ~cancelled_mask
].copy()

after_cancelled = len(clean_df)

print("Rows before:", before_cancelled)
print("Cancelled rows removed:", cancelled_removed)
print("Rows after:", after_cancelled)
print("Customers remaining:", clean_df["CustomerID"].nunique())
```

```python
# Step 4: Remove negative-quantity transactions

before_negative = len(clean_df)

negative_mask = clean_df["Quantity"] < 0
negative_removed = negative_mask.sum()

clean_df = clean_df[
    clean_df["Quantity"] > 0
].copy()

after_negative = len(clean_df)

print("Rows before:", before_negative)
print("Negative-quantity rows removed:", negative_removed)
print("Rows after:", after_negative)
print("Customers remaining:", clean_df["CustomerID"].nunique())
```

```python
# Step 4: Remove negative-quantity transactions

before_negative = len(clean_df)

negative_mask = clean_df["Quantity"] < 0
negative_removed = negative_mask.sum()

clean_df = clean_df[
    clean_df["Quantity"] > 0
].copy()

after_negative = len(clean_df)

print("Rows before:", before_negative)
print("Negative-quantity rows removed:", negative_removed)
print("Rows after:", after_negative)
print("Customers remaining:", clean_df["CustomerID"].nunique())
```

```python
# Step 5: Remove non-positive UnitPrice transactions

before_price = len(clean_df)

invalid_price_mask = clean_df["UnitPrice"] <= 0
invalid_price_removed = invalid_price_mask.sum()

clean_df = clean_df[
    clean_df["UnitPrice"] > 0
].copy()

after_price = len(clean_df)

print("Rows before:", before_price)
print("Non-positive UnitPrice rows removed:", invalid_price_removed)
print("Rows after:", after_price)
print("Customers remaining:", clean_df["CustomerID"].nunique())
```

```python
# Step 6: Remove postage/manual transaction types
# POST = postage
# DOT  = dotcom postage
# M    = manual entries

before_special_codes = len(clean_df)

special_codes = {"POST", "DOT", "M"}

special_code_mask = (
    clean_df["StockCode"]
    .astype(str)
    .isin(special_codes)
)

special_code_removed = special_code_mask.sum()

clean_df = clean_df[
    ~special_code_mask
].copy()

after_special_codes = len(clean_df)

print("Rows before:", before_special_codes)
print("POST/DOT/M rows removed:", special_code_removed)
print("Rows after:", after_special_codes)
print("Customers remaining:", clean_df["CustomerID"].nunique())
```

```python
# Step 7: Final preprocessing validation

print("Final dataset shape:", clean_df.shape)
print()

print("Missing CustomerID:", clean_df["CustomerID"].isna().sum())
print("Duplicate rows:", clean_df.duplicated().sum())
print(
    "Cancelled invoices:",
    clean_df["InvoiceNo"].astype(str).str.startswith("C").sum()
)
print("Negative quantities:", (clean_df["Quantity"] < 0).sum())
print("Zero quantities:", (clean_df["Quantity"] == 0).sum())
print("Zero UnitPrice:", (clean_df["UnitPrice"] == 0).sum())
print("Negative UnitPrice:", (clean_df["UnitPrice"] < 0).sum())
print(
    "POST/DOT/M records:",
    clean_df["StockCode"].astype(str).isin({"POST", "DOT", "M"}).sum()
)
```
